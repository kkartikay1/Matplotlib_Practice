```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=[10,20,30,40]
plt.plot(x,y)
plt.show()
```

```python
# Example ploting a graph with Marker Properties marker ,mfc,mec,ms

x=[10,20,30,40]
y=[10,20,30,40]
plt.plot(x,y,marker="o",mfc="blue",mec="red",ms=20)
plt.show()
```

```python
# Example ploting a graph with Marker Properties marker ,mfc,mec,ms

x=[10,20,30,40]
y=[10,20,30,40]
plt.plot(x,y,marker="d",mfc="blue",mec="red",ms=20)
plt.show()
```

```python
# Example ploting a graph with Marker Properties marker ,mfc,mec,ms

x=[10,20,30,40]
y=[10,20,30,40]
plt.plot(x,y,marker="D",mfc="blue",mec="red",ms=10)
plt.show()
```

```python
# Line Properties of line graph

x=[10,20,30,40]
y=[3,4,5,6]
z=np.array([val+2 for val in y])
plt.plot(x,y,lw=5,c="red")
plt.plot(y,z,ls='dotted',lw=5,c="green")
plt.show()
```

```python
# Line and marker Properties of line graph

x=[10,20,30,40]
y=[3,4,5,6]
z=np.array([val+2 for val in y])
plt.plot(x,y,ls='--',lw=5,c="red",marker="D",mfc="blue",mec="red",ms=10)
plt.plot(y,z,ls='dotted',lw=5,c="green")
plt.show()
```
