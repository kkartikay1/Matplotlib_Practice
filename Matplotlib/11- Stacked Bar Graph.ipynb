```python
# Drawing Stacked Bar Gra[h
year=["2019",'2020','2021','2022']
placement=[150,100,115,125]
cse=[70,40,60,80]
it=[30,20,25,35]
ece=[22,19,20,18]
ece_start=[cse[i]+it[i] for i in range(len(year))]
plt.barh(year,cse,color="red")
plt.barh(year ,it,left=cse,color="blue")
plt.barh(year,ece,left=ece_start,color="green")
plt.show()
```
