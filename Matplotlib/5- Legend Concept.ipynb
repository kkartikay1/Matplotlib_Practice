```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.arange(1,6)
y=x**2
z=x**3
plt.plot(x,y)
plt.plot(x,z)
plt.show()
```

```python
# Apply Lable
```

```python

x=np.arange(1,6)
y=x**2
z=x**3
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend()
plt.show()
```

```python

x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend()
plt.show()
```

```python
# Legend with LOC parameter
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper right")
plt.show()
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="lower center")
plt.show()
```

```python
# framealpha
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper center",framealpha=1)
plt.show()
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper center",framealpha=0.05)
plt.show()
```

```python
# shadow=true
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper center",framealpha=1,facecolor="y",edgecolor="r",shadow=True)
plt.show()
```

```python
# fancybix=true
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper center",framealpha=1,facecolor="y",edgecolor="r",shadow=True,fancybox=True)
plt.show()
```

```python
# fontsize
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
plt.plot(x,k,label="Linear")
plt.plot(x,y,label="SQUARES")
plt.plot(x,z,label="CUBES")
plt.legend(loc="upper center",framealpha=1,facecolor="y",edgecolor="r",
           shadow=True,fancybox=True,fontsize=20)
plt.show()
```

```python
x=np.arange(1,6)
k=np.array([2,6,9,5,3])
y=x**2
z=x**3
lab=("Linear","SQUARES","CUBES")
plt.plot(x,k)
plt.plot(x,y)
plt.plot(x,z)
plt.legend(labels=lab,loc="upper center",framealpha=1,facecolor="y",edgecolor="r",shadow=True,fancybox=True)
plt.show()
```
