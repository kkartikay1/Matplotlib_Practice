```python
# Makes Us to understand x-ticks and y-ticks
x=y=[i for i in range(0,10)]
ax=plt.axes()
# depict illustration
plt.plot(x,y)
# Setting ticks for axis
ax.set_xticks([2,4,6,8,10])
ax.set_yticks([1,3,5,7,9])
plt.show()
```

```python
# Makes Us to understand x-ticks and y-ticks
import math
x=np.arange(0,math.pi*2,0.05)
y=np.sin(x)
ax=plt.axes()

# depict illustration
plt.plot(x,y,color="lime")

# Setting ticks for axis
ax.set_xticks([0,2,4,6])

ax.set_yticks([-1,0,1])
plt.show()
```

```python
# Makes Us to understand x-ticks and y-ticks
import math
# assign coordinates
x=np.arange(0,math.pi*2,0.05)
y=np.sin(x)
ax=plt.axes()

plt.xlabel("x-axis")
plt.ylabel("y-axis")

# depict illustration
plt.plot(x,y,color="lime")

# Setting ticks for axis
ax.set_xticks([0,2,4,6])

ax.set_yticks([-1,0,1])

# Setting labes for y tick
ax.set_yticklabels(["sin(-90deg)","sin(0deg)","sin(90deg)"])
plt.show()
```

```python
# Makes Us to understand x-ticks and y-ticks
import math
# assign coordinates
x=np.arange(0,math.pi*2,0.05)
y=np.sin(x)
ax=plt.axes()

# depict illustration
plt.plot(x,y,color="lime")

ax.set_xlabel("angle")
ax.set_ylabel("sine")

# Setting ticks for axis
ax.set_xticks([0,2,4,6])

# Setting labes for X tick
ax.set_xticklabels(["Zero","two","four","six"])

ax.set_yticks([-1,0,1])

plt.show()
```
