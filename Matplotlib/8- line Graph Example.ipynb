```python
import matplotlib.pyplot as plt
day=["MON","TUE","WED","THU","FRI","SAT","SUN"]
sleep=[7,6,5,8,9,9,5]
plt.plot(day,sleep)
plt.show()
```

```python
import matplotlib.pyplot as plt
day=["MON","TUE","WED","THU","FRI","SAT","SUN"]
sleep=[7,6,5,8,9,9,5]
plt.plot(day,sleep,marker="d",mfc="r",mec="b",ms=10,color="green")
plt.show()
```
